```python
# Exercise 1
ENGINEER = "ALCORIZA"
STUDENT_ID = "TUPM-26-1085"
SEED_NUM = int(STUDENT_ID[-1])

print(f"System Initialized for: {ENGINEER}")
print(f"Algorithmic Seed Value: {SEED_NUM}")

# 2 sensor data
sensor_data = []

for letter in ENGINEER:
    value = ord(letter) + SEED_NUM
    sensor_data.append(value)

# fixed number sensor readings
sensor_data = sensor_data[:5]

print("Generated Sensor Data:", sensor_data)

# 3. operating limit
limit = (SEED_NUM * 10) + 20

print("Safety Limit:", limit)

# 4. receiving and validating the readings
valid_readings = []
invalid_readings = []
classification_results = []

for i in range(5):

    user_input = input("Enter Sensor Reading " + str(i + 1) + ": ")

    try:
        reading = int(user_input)

        # Validate the reading
        if reading < 0:
            print("Invalid Reading")
            invalid_readings.append(user_input)

        else:
            print("Valid Reading")
            valid_readings.append(reading)

              # 5. CLASSIFY READINGS
            if reading < 5:
                result = "LOW SIGNAL"

            elif reading <= limit:
                result = "NOMINAL"

            else:
                result = "OVER LIMIT"

            print("Classification:", result)
            classification_results.append(result)

    except ValueError:
        print("Invalid Input: Please enter a number.")
        invalid_readings.append(user_input)

        # 6. final step
print("------ FINAL OUTPUT ------")

print("Generated Sensor Data:", sensor_data)
print("Valid Results:", valid_readings)
print("Invalid Results:", invalid_readings)
print("Classification Results:", classification_results)

print("Total Readings:", 5)
print("Total Valid:", len(valid_readings))
print("Total Invalid:", len(invalid_readings))
print("Safety Limit:", limit)

print("Execution Log:")
if len(valid_readings) + len(invalid_readings) == 5:
    print("all readings processed successfully.")
else:
    print("Some readings were not processed correctly.")
```

```python
#  Alcoriza Exercise 2

ENGINEER = "ALCORIZA"
STUDENT_ID = "TUPM-26-1085"
SEED_NUM = int(STUDENT_ID[-1])

print("System Initialized for:", ENGINEER)
print("Seed Number:", SEED_NUM)


# 2. signal 
signal = ENGINEER + " " + " " + str(SEED_NUM)

print("Generated Signal:", signal)


# 3. process signal
processed_signal = signal.upper()

print("Processed Signal:", processed_signal)


# 4. analuze characters
letters = 0
numbers = 0
spaces = 0
others = 0

print("------ CHARACTER ANALYSIS ------")

for character in processed_signal:

    if character.isalpha():
        print(character, "- LETTER")
        letters += 1

    elif character.isdigit():
        print(character, "- NUMBER")
        numbers += 1

    elif character.isspace():
        print(character, "- SPACE")
        spaces += 1

    else:
        print(character, "- OTHER")
        others += 1


# 5. classification
if others > 0:
    classification = "this contains other classification"

elif numbers > 0:
    classification = "letters and numbers"

else:
    classification = "letter"


# 6. DIAGNOSTIC REPORT
print("===== DIAGNOSTIC REPORT =====")
print("Generated Signal:", signal)
print("Processed Signal:", processed_signal)

print("Character Analysis:")
print("Letters:", letters)
print("Numbers:", numbers)
print("Spaces:", spaces)
print("Other Characters:", others)

print("Signal Classification:", classification)

print("Execution Log:")
if len(valid_readings) + len(invalid_readings) == 5:
    print("all readings processed successfully.")
else:
    print("some readings were not processed correctly.")
```

```python
# Alcoriza Exercise 3
ENGINEER = "ALCORIZA"
STUDENT_ID = "TUPM-26-1085"
SEED_NUM = int(STUDENT_ID[-1])

print("System Initialized for:", ENGINEER)
print("Seed Number:", SEED_NUM)

# 2. signal 
password = ENGINEER + str(SEED_NUM)

# Normalize password
password = password.upper()

print("Generated Password:", password)

# 3. process signal
attempt_limit = SEED_NUM + 2

print("Attempt Limit:", attempt_limit)


# 4 authentication process
attempts = 0
access_granted = False
while attempts < attempt_limit:

    user_password = input("Enter Password: ")

    user_password = user_password.strip().upper()

    attempts += 1

    if user_password == password:
        print("Access Granted")
        access_granted = True
        break

    else:
        print("Incorrect Password.")
        print("Attempts Made:", attempts)


# 5. system
if access_granted:
    access_result = "ACCESS GRANTED"
    system_state = "UNLOCKED"

else:
    access_result = "ACCESS DENIED"
    system_state = "LOCKED"

print("Execution Log:")
if access_granted:
    print("Access granted on attempt", attempts)
else:
    print("Access denied after", attempts, "attempts")

# 6. FINAL OUTPUT
print("===== FINAL OUTPUT =====")
print("Generated Password:", password)
print("Attempt Limit:", attempt_limit)
print("Attempts Made:", attempts)
print("Access Result:", access_result)
print("Final System State:", system_state)

print("Execution Log:")
if access_granted:
    print("  Access granted on attempt", attempts)
else:
    print("  Access denied after", attempts, "attempts")
```
